```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
```

```python
data = pd.read_csv('spending.csv')
data.head()
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(data)
```

```python
dbscan = DBSCAN(eps=0.5, min_samples=5, metric='euclidean', algorithm='auto', leaf_size=30, p=None, n_jobs=1)
```

```python
dbscan.fit(x_scaled)
```

```python
dbscan.cluster_centers_
```

```python
data["Cluster"] = dbscan.labels_
data.head()
```

```python
plt.scatter(x_scaled[:,1],x_scaled[:,2],c=dbscan.labels_,cmap='rainbow')
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.show()
```
